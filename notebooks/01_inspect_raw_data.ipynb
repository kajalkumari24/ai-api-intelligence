```python
import numpy as np

def encode_metadata(df, auth_categories):
    auth_dummies = pd.get_dummies(df['auth'], prefix='auth')
    auth_dummies = auth_dummies.reindex(columns=auth_categories, fill_value=0)
    https_col = df['https'].astype(int)
    cors_col = df['cors'].map({True: 1, False: 0}).fillna(0.5)
    return pd.concat([https_col, cors_col, auth_dummies], axis=1).values

# Learn the full set of auth categories from TRAIN only (same rule as TF-IDF fit)
auth_categories = pd.get_dummies(train_df['auth'], prefix='auth').columns

meta_train = encode_metadata(train_df, auth_categories)
meta_val = encode_metadata(val_df, auth_categories)
meta_test = encode_metadata(test_df, auth_categories)

X_train_combined = np.hstack([X_train_emb2, meta_train])
X_val_combined = np.hstack([X_val_emb2, meta_val])
X_test_combined = np.hstack([X_test_emb2, meta_test])

print(X_train_combined.shape, X_val_combined.shape, X_test_combined.shape)
```

```python
from transformers import DistilBertTokenizer

tokenizer = DistilBertTokenizer.from_pretrained('distilbert-base-uncased')

def tokenize_texts(texts, max_length=128):
    return tokenizer(
        texts, padding='max_length', truncation=True,
        max_length=max_length, return_tensors='pt'
    )

train_encodings = tokenize_texts(train_df['text_features'].tolist())
val_encodings = tokenize_texts(val_df['text_features'].tolist())
test_encodings = tokenize_texts(test_df['text_features'].tolist())

print(train_encodings['input_ids'].shape)
```

```python
import torch

class APIDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        item = {key: val[idx] for key, val in self.encodings.items()}
        item['labels'] = torch.tensor(self.labels[idx])
        return item

train_dataset = APIDataset(train_encodings, train_df['label'].tolist())
val_dataset = APIDataset(val_encodings, val_df['label'].tolist())
test_dataset = APIDataset(test_encodings, test_df['label'].tolist())

print(len(train_dataset), len(val_dataset), len(test_dataset))
```
